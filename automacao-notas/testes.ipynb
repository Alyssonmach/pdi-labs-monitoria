```python
from tkinter import filedialog
import tkinter as tk
import re
import os
```

```python
NAME_REFERENCE = 'Aluno'
QUESTION_WEIGHTS_REFERENCE = 'Taxa de acerto'
```

```python
def ask_for_directory():

    root = tk.Tk()
    root.withdraw()

    return filedialog.askdirectory()
```

```python
def load_jupyter(path_file):

    data = open(path_file, mode = 'r', encoding = 'utf-8')
    jupyter_file = data.read()
    data.close()

    return jupyter_file
```

```python
def find_all_occurrences_index(string, sub_string):

    return [sub_init.start() for sub_init in re.finditer(sub_string, string)]
```

```python
def get_values(string, list_index, max_lenght):

    values = list()
    for index in list_index:
        values.append(string[index:(index + max_lenght)])
    
    return values
```

```python
def get_files_in_dir(path_dir, extension, excesion = ''):

    list_files, ipynb_files = os.listdir(path_dir), list()
    for files in list_files:
        if files.find(extension) != -1 and files.find(excesion) <= 0: 
            ipynb_files.append(path_dir + '/' + files)

    return ipynb_files 
```

```python
def adjust_names(name_string):
    name = name_string.split(':')[1].split('"')[0]
    if name[0].isspace(): name = name[1:]

    return name
```

```python
def adjust_weight(weight_string):
    
    weight = [int(word) for word in list(weight_string) if word.isdigit()]
    if len(weight) > 2 or len(weight) <= 0: weight = ''
    elif len(weight) == 2: weight = int(str(weight[0]) + str(weight[1])) / 10
    else: weight = weight[0]

    if weight > 1.0 or weight < 0.0: weight = ''

    return weight 
```

```python
def extract_information(path_file):

    jupyter_file = load_jupyter(path_file = path_file)
    names = find_all_occurrences_index(string = jupyter_file, sub_string = NAME_REFERENCE)
    weights = find_all_occurrences_index(string = jupyter_file, sub_string = QUESTION_WEIGHTS_REFERENCE)
    names = get_values(string = jupyter_file, list_index = names, max_lenght = 100)
    weights = get_values(string = jupyter_file, list_index = weights, max_lenght = 30)

    return list(map(adjust_names, names)), list(map(adjust_weight, weights))
```

```python
path_files = 'C:/Users/Alysson/OneDrive/Documents/projetos-github/pdi-laboratories/laboratorios-2021-1/laboratorio02'
```

```python
files = get_files_in_dir(path_dir = path_files, extension = '.ipynb', excesion = 'checkpoints')
results = map(extract_information, files)
name, weights = list(results)[0]
print(name, weights)
```
